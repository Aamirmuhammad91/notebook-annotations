An [increase](https://openenergyplatform.org/ontology/oeo/OEO_00340005) in the magnetic field strength is relevant for the [efficiency](https://openenergyplatform.org/ontology/oeo/OEO_00140050) of fusion [power](https://openenergyplatform.org/ontology/oeo/OEO_00000333) experiments.

[Geothermal energy](https://openenergyplatform.org/ontology/oeo/OEO_00000191) is extracted from the Earth's interior to [power](https://openenergyplatform.org/ontology/oeo/OEO_00000333) a nearby [district heating](https://openenergyplatform.org/ontology/oeo/OEO_00000132) [system](http://purl.obolibrary.org/obo/RO_0002577).

The [petroleum](https://openenergyplatform.org/ontology/oeo/OEO_00000115) [reservoir](https://openenergyplatform.org/ontology/oeo/OEO_00010084) contains a significant [amount](https://openenergyplatform.org/ontology/oeo/OEO_00340069) of [crude oil](https://openenergyplatform.org/ontology/oeo/OEO_00000115), which is a [fossil energy](https://openenergyplatform.org/ontology/oeo/OEO_00020149) resource.

The [wind turbine](https://openenergyplatform.org/ontology/oeo/OEO_00000044) converts the [kinetic energy](https://openenergyplatform.org/ontology/oeo/OEO_00230020) of [wind](https://openenergyplatform.org/ontology/oeo/OEO_00000043) into mechanical [energy](https://openenergyplatform.org/ontology/oeo/OEO_00000150).

The [photovoltaic power plant](https://openenergyplatform.org/ontology/oeo/OEO_00000324) generates electric [energy](https://openenergyplatform.org/ontology/oeo/OEO_00000150) using [solar radiation](https://openenergyplatform.org/ontology/oeo/OEO_00020038).
