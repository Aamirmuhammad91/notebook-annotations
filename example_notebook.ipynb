[Electrical energy](https://openenergyplatform.org/ontology/oeo/OEO_00000139) is a form of [energy](https://openenergyplatform.org/ontology/oeo/OEO_00000150) derived from the [potential](https://openenergyplatform.org/ontology/oeo/OEO_00290000) or [kinetic energy](https://openenergyplatform.org/ontology/oeo/OEO_00230020) of charged particles.

[Wind turbine](https://openenergyplatform.org/ontology/oeo/OEO_00000044)s convert [kinetic energy](https://openenergyplatform.org/ontology/oeo/OEO_00230020) of moving [air](https://openenergyplatform.org/ontology/oeo/OEO_00000054) into [electrical energy](https://openenergyplatform.org/ontology/oeo/OEO_00000139).
