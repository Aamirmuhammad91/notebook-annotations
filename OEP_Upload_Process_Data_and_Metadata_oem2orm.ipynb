```python
db = oem2orm.setup_db_connection()
```

### Creating sql tables from oe[metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119)

The oe[metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) format is a standardised json file format and required for all [data](http://purl.obolibrary.org/obo/IAO_0000027) uploaded to the OEP. It includes the [data](http://purl.obolibrary.org/obo/IAO_0000027) [model](https://openenergyplatform.org/ontology/oeo/OEO_00020352) and the used [data](http://purl.obolibrary.org/obo/IAO_0000027) types. This allows us to derive the necessary tables in - SQLAlchemy from it.

In order to create the table(s) we need to tell python where to find our oe[metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) file first. To do this we place them in the folder "[metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119)" which is in the current directory (Path of this jupyter notebbok). Provide the path to your own folder if you want to use your own [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119). oem2orm will [process](http://purl.obolibrary.org/obo/BFO_0000015) all files that are located in the folder.


```python
metadata_folder = oem2orm.select_oem_dir(oem_folder_name="metadata")
```

The next command will set up the table. The collect_tables_function collects all [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) files in a folder and retrives the SQLAlchemy ORM [object](http://purl.obolibrary.org/obo/BFO_0000030)s and returns them. The Tables are ordered by foreign key. Having a valid [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) strings is necessary for the following steps.


<div class="alert alert-block alert-info">
INFO: The red output is information printed by the logger. It does not mean that an error has occurred.</div>


```python
tables_orm = oem2orm.collect_tables_from_oem(db, metadata_folder)
```

Now we can use create our table [object](http://purl.obolibrary.org/obo/BFO_0000030)s in the [database](https://openenergyplatform.org/ontology/oeo/OEO_00000118).

```python
#create table
oem2orm.create_tables(db, tables_orm)
```

The tables should now be public, but empty on the OEP at the location provided in the [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) file. For this example tutorial, the created table is located in [model_draft.upload_tutorial_example_[data](http://purl.obolibrary.org/obo/IAO_0000027)](https://openenergyplatform.org/[data](http://purl.obolibrary.org/obo/IAO_0000027)edit/view/model_draft/upload_tutorial_example_[data](http://purl.obolibrary.org/obo/IAO_0000027)). If you've just been playing around and don't want to write any [data](http://purl.obolibrary.org/obo/IAO_0000027) to the OEP, please make sure to delete your tables again. 

### Deleting your table

Running the following commands will delete the tables from the [database](https://openenergyplatform.org/ontology/oeo/OEO_00000118) which you have in your ordered ORM. Take care that you only delete tables you actually want to have removed.

<div class="alert alert-block alert-danger">
Skip this command, if you want to keep your table!

```python
# In order to actually delete, you will need to type: yes
oem2orm.delete_tables(db, tables_orm)
```

## Writing [data](http://purl.obolibrary.org/obo/IAO_0000027) into a table

In this example we will upload [data](http://purl.obolibrary.org/obo/IAO_0000027) from a [csv](https://openenergyplatform.org/ontology/oeo/OEO_00280003) file. Pandas has a read_[csv](https://openenergyplatform.org/ontology/oeo/OEO_00280003) [function](http://purl.obolibrary.org/obo/BFO_0000034) which makes importing a [csv](https://openenergyplatform.org/ontology/oeo/OEO_00280003)-file rather comfortable. It reads [csv](https://openenergyplatform.org/ontology/oeo/OEO_00280003) into a [Data](http://purl.obolibrary.org/obo/IAO_0000027)Frame. By default, it assumes that the fields are comma-separated. Our example file has columns with semicolons as separators, so we have to specify this when reading the file.

The example file for this tutorial ('upload_tutorial_example_[data](http://purl.obolibrary.org/obo/IAO_0000027).[csv](https://openenergyplatform.org/ontology/oeo/OEO_00280003)') is in the '[data](http://purl.obolibrary.org/obo/IAO_0000027)' directory, next to this tutorial. Make sure to adapt the path to the file you're using if your file is located elsewhere.

```python
# db = oem2orm.setup_db_connection()

filepath = "./data/upload_tutorial_example_data.csv"
example_df = pd.read_csv(filepath, encoding='utf8', sep=';')

# show the first 10 row´s
example_df[:10]
```

We need to define the location in the OEDB where the [data](http://purl.obolibrary.org/obo/IAO_0000027) should be written to. The connection information is still available from our steps above. 

```python
# change to the table schema
schema = "sandbox"
# change to the table name
table_name = "upload_tutorial_example_data"
connection = db.engine
```

The following command will write the content of your dataframe to the table on the OEP that was created earlier. <br> The table is uploaded in chunks to ensure a stable connection to the OEP when uploading large tables.<br>
Take a look at your table on the OEP after succesfull upload or for debugging in case of an exception!

```python
# Define chunksize. The chunksize defines the number of uploaded csv-rows per request.
chunksize = 6000

with pd.read_csv(filepath, encoding='utf8', sep=';', chunksize=chunksize) as reader:
    for chunk in reader:
        print(f'Uploading chunk from table: {table_name}.')
        try: 
            chunk.to_sql(table_name, connection, schema=schema, if_exists='append', index=False)
            print(f'Appended {len(chunk)} rows to table on OEP')
        except Exception as e:
            print(f'Writing to {schema}.{table_name} failed!')
            print('Delete and recreate with the commands above, if you want to repeat your upload again.')
            print('The program throws the following exception:')
            print(e)
        finally:
            print(f'Take a look at your table on the OEP: https://openenergyplatform.org/dataedit/view/{schema}/{table_name}')
```

## Writing [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) to the table

Now that we have [data](http://purl.obolibrary.org/obo/IAO_0000027) in our table it's high time, that we attach our [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) to it. Since we're using the [api](https://openenergyplatform.org/ontology/oeo/OEO_00000059), some direct http-requests and a little helper [function](http://purl.obolibrary.org/obo/BFO_0000034) from the oep-client, we need to [import](https://openenergyplatform.org/ontology/oeo/OEO_00020201) these new dependencies. 

```python
import json
import requests
```

We use the [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) folder we set up before. (See the [Creating tables section](https://openenergyplatform.github.io/academy/tutorials/99_other/OEP_Upload_Process_Data_and_[Metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119)_oem2orm/#creating-sql-tables-from-oe[metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119))
If you want to set another folder use the code below:

```python
# oem_path = oem2orm.select_oem_dir(oem_folder_name="metadata")
md_file_name = "example_metadata"
```

First we're reading the [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119) file into a json dictionary.

```python
metadata = oem2orm.mdToDict(oem_folder_path=metadata_folder, file_name=md_file_name)
```

Then we need to validate the [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119).

```python
oem2orm.omi_validateMd(metadata)
```

Now we can upload the [metadata](https://openenergyplatform.org/ontology/oeo/OEO_00000119).

```python
oem2orm.api_updateMdOnTable(metadata)
```
